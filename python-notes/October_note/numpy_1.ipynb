```python
import numpy as np
np.random.seed(0)
```

python作为一种编译型语言，虽然十分简洁易懂，但执行效率偏低  
例如列表，由于列表并不指定其内部元素的类型，导致在列表执行操作时会优先判断元素的type，导致执行效率大打折扣  
于是，专为数值运算开发的numpy应运而生

作为numpy的基本元素：数组array有以下几种生成方式

1. 从 Python 列表/元组等序列创建

```python
# 从列表创建
a = np.array([1, 2, 3])

b = np.array([[1, 2], 
              [3, 4]])

print(a,"\n")
print(b,"\n")

# 从元组创建
c = np.array((1, 2, 3))
print(c)
```

2. 创建全 0、全 1 或空数组

```python
# 全 0 数组
zeros = np.zeros((2, 3))          # shape 为 (2,3)
zeros_int = np.zeros(5, dtype=int)
print(zeros,"\n")
print(zeros_int,"\n")
# 全 1 数组
ones = np.ones((2, 2))
print(ones)
# 未初始化的“空”数组（含随机垃圾值）
empty = np.empty((3, 3))
```

3. 创建特定数值填充的数组

```python
# 填充指定值
full = np.full((2, 4), 7)         # 用 7 填充 2x4 数组
print(full,"\n")

# 创建单位矩阵（对角线为1，其余为0）
eye = np.eye(3)                   # 3x3 单位矩阵
identity = np.identity(3)         # 同上
print(eye,"\n")
print(identity)
```

4. 创建等差或等比数列

```python
# 等差数列
arange = np.arange(0, 10, 2)      # [0, 2, 4, 6, 8]
print(arange,"\n")

# 在指定区间内生成固定数量的点（包含终点）
linspace = np.linspace(0, 1, 5)   # [0. , 0.25, 0.5 , 0.75, 1. ]
print(linspace,"\n")
# 等比数列（对数间隔）
logspace = np.logspace(0, 2, 5)   # 10^0 到 10^2 之间 5 个点
print(logspace)
```

5. 特殊数组（随机、对角等）

```python
# 随机数组
rand = np.random.rand(2, 3)           # [0,1) 均匀分布
print(rand,"\n")
randn = np.random.randn(2, 3)         # 标准正态分布
print(randn,"\n")
random_int = np.random.randint(0, 10, size=(2, 2))     # 随机整数
print(random_int,"\n")
# 对角矩阵
diag = np.diag([1, 2, 3])             # 创建对角阵
print(diag,"\n")
extract_diag = np.diag([[1,2],
                        [3,4]]) # 提取对角线元素 → [1,4]

print(extract_diag)
```

数组的拼接

```python
x = np.array([1, 2, 3])
y = np.array([3, 2, 1])
z = np.array([5, 6, 7])
np.concatenate([x, y,z])
```

沿着固定维度处理数组时，使用 np.vstack（垂直栈）和 np.hstack（水平栈）函数会更简洁：

```python
x = np.array([1, 2, 3])
grid = np.array([[9, 8, 7],
                 [6, 5, 4]])
print(np.vstack([x, grid]))
y = np.array([[99],
              [99]])
print(np.hstack([grid, y]))
```

相关的 np.hsplit和 np.vsplit的用法也类似：

```python
grid = np.arange(16).reshape((4, 4))
print(grid,"\n")
upper, lower = np.vsplit(grid, [1])
print(upper,"\n")
print(lower,"\n")
left, right = np.hsplit(grid, [2])
print(left,"\n")
print(right)
```

数组的计算：广播

```python
import numpy as np
a = np.array([0, 1, 2])
b = np.array([5, 5, 5])

M = np.ones((3, 3))
print(M + a)
```

广播规则  
1：如果两个数组的维度数不相同，那么小维度数组的形状将会在最左边补 1  
2：如果两个数组的形状在任何一个维度上都不匹配，那么数组的形状会沿着维度为 1 的维度扩展以匹配另外一个数组的形状。  
3：如果两个数组的形状在任何一个维度上都不匹配并且没有任何一个维度等于 1，那么会引发异常。  

```python
M = np.ones((2, 3))
a = np.arange(3)
print(M, a)
print(M + a)
```

```python
a = np.arange(3).reshape((3, 1))
b = np.arange(3)
print(a, b)
print(a + b)
```
