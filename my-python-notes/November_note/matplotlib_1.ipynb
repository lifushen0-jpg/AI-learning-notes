```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

#创建图表并设置大小
plt.figure(figsize = (10,5))

# ====================================================================================================
# ====================================================================================================

#图表数据
month = ['1月', '2月', '3月', '4月']
sales = [100, 40 ,120, 60]

#绘制图表(折线图)
plt.plot(month,sales
         ,label = '产品A'   
         ,marker = "o")

#添加坐标轴标签
plt.xlabel('month')
plt.ylabel('销售额')

#添加图例
plt.legend(loc = 'upper left')
plt.grid(True, alpha = 0.1, color = 'blue', linestyle = '--')  #plt.grid(axis = 'x')  # 

# 设置刻度字体大小
plt.xticks(rotation = 45, fontsize = 10)

# 设置Y轴范围
plt.ylim(0,160)

# 在每个数据点上标记数值
for x, y in zip(month, sales):
    plt.text(x, y+5, str(y), ha = "center", va = "bottom")

# ====================================================================================================
# ====================================================================================================

#添加标题
plt.title('114514',color = 'red', fontsize = 20)

#显示图表
plt.show()
```

## 条形图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

# 创建图表并设置大小
plt.figure(figsize = (10,5))

# ====================================================================================================
# ====================================================================================================

# 图表数据
subject = ['Chinese', 'English', 'mathematica', 'science']
score = [85, 92, 78, 88]

plt.bar(subject, score,
        label = 'My',
        color = 'orange',
        width = 0.6)

plt.xlabel('subject',fontsize = 10)
plt.ylabel('score', fontsize = 10)

plt.grid(axis="y", alpha = 0.5, color = 'blue', linestyle = '--')  #plt.grid(axis = 'x')  # 


# ====================================================================================================
# ====================================================================================================

# 自动优化排版(效果较差)
plt.tight_layout()

# 添加标题
plt.title("My scores",color = 'red', fontsize = 20)

# 显示图表
plt.show()
```

# 条形图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

# 创建图表并设置大小
plt.figure(figsize = (10,5))

# ====================================================================================================
# ====================================================================================================

# 图表数据
subject = ['Chinese', 'English', 'mathematica', 'science']
score = [85, 92, 78, 88]

plt.barh(subject, score,
        label = 'My',
        color = 'orange',
        height = 0.6)

plt.xlabel('subject',fontsize = 10)
plt.ylabel('score', fontsize = 10)

plt.grid(axis="x", alpha = 0.5, color = 'blue', linestyle = '--')  #plt.grid(axis = 'x')  # 


# ====================================================================================================
# ====================================================================================================

# 自动优化排版(效果较差)
plt.tight_layout()

# 添加标题
plt.title("My scores",color = 'red', fontsize = 20)

# 显示图表
plt.show()
```

# 饼图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

# 创建图表并设置大小
plt.figure(figsize = (10,5))

# ====================================================================================================
# ====================================================================================================

# 图表数据
things = ['learn', 'entertainment', 'sport', 'sleep', 'others']
times = [9, 3, 2, 8 ,2]
colors = ['red', '#99ff99', '#ffcc99', "#263892", '#456678']

plt.pie(times, labels=things,
        autopct ="%.2f%%",
        startangle=0,
        colors = colors)


# ====================================================================================================
# ====================================================================================================

# 自动优化排版(效果较差)
plt.tight_layout()

# 添加标题
plt.title('how much time do I spend in the different things', fontsize = 10, color = "blue")

# 显示图表
plt.show()
```

# 环形图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

# 创建图表并设置大小
plt.figure(figsize = (10,5))

# ====================================================================================================
# ====================================================================================================

# 图表数据
things = ['learn', 'entertainment', 'sport', 'sleep', 'others']
times = [9, 3, 2, 8 ,2]
colors = ['red', '#99ff99', '#ffcc99', "#263892", '#456678']

plt.pie(times, labels=things,
        autopct ="%.2f%%",
        startangle=0,
        colors = colors,
        wedgeprops={'width':0.6},    # 圆环大小
        pctdistance=0.6              # 数据到中心的距离
        )        


# ====================================================================================================
# ====================================================================================================

# 自动优化排版(效果较差)
plt.tight_layout()

# 添加标题
plt.title('how much time do I spend in the different things', fontsize = 10, color = "blue")

# 显示图表
plt.show()
```

# 爆炸式饼图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

# 创建图表并设置大小
plt.figure(figsize = (10,5))

# ====================================================================================================
# ====================================================================================================

# 图表数据
things = ['learn', 'entertainment', 'sport', 'sleep', 'others']
times = [9, 3, 2, 8 ,2]
colors = ['red', "#35617e", '#ffcc99', "#268B92", "#102186"]

plt.pie(times, labels=things,
        autopct ="%.2f%%",
        startangle=0,
        colors = colors,
        explode= [0.05, 0, 0, 0, 0],
        shadow=True
        )


# ====================================================================================================
# ====================================================================================================

# 自动优化排版(效果较差)
plt.tight_layout()

# 添加标题
plt.title('how much time do I spend in the different things', fontsize = 10, color = "blue")

# 显示图表
plt.show()
```
