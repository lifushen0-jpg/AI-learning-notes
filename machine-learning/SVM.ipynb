# 支持向量机（SVM）
### 分类：线性分类
### 与感知机的区别：SVM 要求分界线与数据的间隔的最小值最大，感知机无此要求

下式为距离函数：

$$
γ_i = y_i ( (w / ||w||)·x_i + b / ||w|| ) = (γ̂_i) / ||w||
$$

```python
import torch

def distance_to_hyperplane(x, y, w, b):
    """
    计算每个样本到超平面 w^T x + b = 0 的带符号距离（乘以标签 y）
    
    参数:
        x: [n_samples, n_features]
        y: [n_samples]
        w: [n_features]
        b: scalar (float or tensor)
    
    返回:
        [n_samples] 的张量：每个样本的 y_i * (w^T x_i + b) / ||w||
    """
    linear_output = x @ w + b  
    
    w_norm = torch.norm(w)  
    
    signed_distances = y * linear_output / w_norm
    
    return signed_distances

x = torch.tensor([[1., 1.],
                  [2., 2.]])      
y = torch.tensor([1., -1.])       
w = torch.tensor([1., -1.])       
b = torch.tensor(1.0)        

result = distance_to_hyperplane(x, y, w, b)
print(result)
```
