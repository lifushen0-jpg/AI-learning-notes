# 三、判断与循环

## 1.判断句：if

```python
# 语法： if 判断句（通常返回布尔值） ：
# 当判断句为真时执行下一行内容
if 2 == 1 + 1:
    print(2)

if 2 == 1:   
    print(5) 


# 当if后面为非零非空数据时也会执行下一行
if 1:
    print(1)

if 0:
    print(0)

if None:
    print(0)

if "114":
    print(114) 
    
if "":
    print("空")
```

```python
# if-else语句：当if执行失败时直接执行else，若if执行成功则else不会执行
if 1 == 1:
    print(114)
else:
    print(514)


if 1 == 2:
    print("2025")
else:
    print("Nov")
```

```python
# if-elif-else结构
# 它依次检查每个条件测试，直到遇到通过了的条件测试。测试通过后，Python将执行紧跟在它后面的代码，并跳过余下的测试。
age = 12
if age < 4:
    print("Your admission cost is $0.")
elif age < 18 and age >= 4:
    print("Your admission cost is $25.")
else:
    print("Your admission cost is $40.")
```

## 2.for循环（遍历循环）
当你需要 **遍历** 一个数据集合（如列表、字典、字符串）中的每一个元素时，就用 `for` 循环。

### 语法：for 变量 in 范围:
 范围可为暂存数据的结构（见下文）和 range 等

#### **`for` 循环的好帮手: `range()`**

当你需要重复执行一段代码 **固定的次数** 时，使用 `range()`。
使用 range(x) 函数，就可以生成一个从 0 到 x-1 的整数序列。
如果是 `range(a,b)`  函数，你可以生成了一个左闭右开的整数序列。

```python
# range(起始值，终止值，步长)  # 取值范围为 [ 起始值，终止值 )
for x in range(1,11,2):
    print(x, end = " ")

for x in range(1,12):  # 只填入两个值是默认步长为1
    print(x, end = "")
    
for x in range(12):   # 只填入一个值时默认步长为1，初始值为0
    print(x, end = "")

count = 0
for x in range(100):
    count += x
print("\n",count)
```

## 3.while 循环（条件循环）

```python
#语法： while 判断句：   当判断句为真时执行后面的语句（对“真”的判断与if相同）
n = 100
count =0
while n > 0:
    count += n
    n -= 1
print(count)
```

## break 与 continue


```python
#两者都能在 for 和 while 循环中运行
#break会终止当前所在循环
count = 0
for x in range(1,101):
    if x == 51:
        break
    count += x
print(count)
```

```python
#continue 会跳过此次循环
count = 0
for x in range(1,101):
    if x == 51:
        continue
    count += x
print(count)


n = 100
count =0
while n > 0:
    if n == 50:
        n -= 1
        continue
    count += n
    n -= 1
print(count)
```

### **3. 元组 (Tuple) **

- **元组 (Tuple)**: "不可变"的列表。
    - 用 `()` 定义，单个元素时需加逗号，例如 `(1,)`。
    - 一旦创建，就不能再添加、修改或删除元素。

```python
# 元组常用于保护数据，因为它无法被修改
tuple = (2,"1") 
print(tuple)
print(type(tuple))
```

```python
#元组中只有一个数据时，末尾加逗号，否则保存为该数据。
tuple = ("2")
print(type(tuple))
tuple = (2,)
print(type(tuple))
```

```python
def PrimeNumber_judge_1(n):  # ()放要传入的参数
    for i in range (2,int(n**0.5)+1):
        if n%i==0:
            print("这不是个素数")
            return   # 返回return后面的值并终止函数运行
    print("这是个素数")

def PrimeNumber_judge_2(n):
    for i in range (2,int(n**0.5)+1):
        if n%i==0:
            return "这不是个素数"
    return "这是个素数"

PrimeNumber_judge_1(11451)
print(PrimeNumber_judge_2(97))
print(PrimeNumber_judge_2(100))
```

```python
# 函数嵌套
# 嵌套调用
def outer(a,b):
    c = b*b + inner(a)
    print(c)
def inner(a):
    return a**a
outer(3, 2)

# 嵌套定义
def outer(a,b):
    def inner(a):
        return a**a
    c = b*b + inner(a)
    print(c)
outer(3, 2)

print(inner(4))
```

```python
# 作用域
c = 9   # 这是个全局变量
def func3(a): 
    b = 3    # 这是个局域变量
    print(a + b + c)

def func4():
    global b    # 把局部变量转变成全局变量
    b = 1

func3(4)
```

```python
# 匿名函数
# 语法：函数名 = lambda  形参:返回值（表达式）
Sum = lambda a, b : a + b
print(Sum(1, 2))  #函数名（实参）
```

```python
# 闭包
# 条件：   
# 1，函数嵌套定义    2.内层函数使用外层函数的局部变量    3.外层函数的函数的返回值是内层函数的函数名
def outer(m):
    n=10
    def inner(w):
        print(n+m+w)
    return inner

print(outer(1))
outer(1)(2)
```

```python
#一个应用：计数器

def make_counter():
    count = 0  # 外层函数的局部变量
    def counter():
        nonlocal count  # 引用外层变量
        count += 1
        return count
    return counter  # 返回内层函数

c = make_counter()
print(c())  # 1（记住了count的状态）

for i in range(10):
    print(c())
```
