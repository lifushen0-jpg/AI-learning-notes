节省内存

```python
# 幸运的是，执行原地操作非常简单。我们可以使用切片表示法将操作的结果分配给先前分配的数组
Z = torch.zeros_like(Y)
print('id(Z):', id(Z))
Z[:] = X + Y
print('id(Z):', id(Z))

before = id(X)
X += Y
print(id(X) == before)

before = id(X)
X[:] = X + Y
id(X) == before
```

```python
A = X.numpy()
B = torch.tensor(A)
type(A), type(B)
```
