DataFrame学习笔记

```python
import numpy as np
import pandas as pd
```

1. Creat a DataFrame

```python
#Creat by Series
s1 = pd.Series([1 ,2 ,3 ,4 ,5])
s2 = pd.Series([6, 7, 8, 9, 10])
df = pd.DataFrame({"List_1": s1, "List_2": s2})
print(df)
print(type(df), type(df["List_1"]))

#Creat by dict
df = pd.DataFrame(
    {
        "id": [1,2,3],
        "name": ["Tom", "Jack", "Bob"],
        "age": [15, 14, 15.5]
    }, index= [1, 2, 3], columns= ["name", "age", "id"]
)
print(df)
```

2. DataFrame的属性：  
index.(), values(), dtypes(), shape(), ndim()#维度, size()#元素个数, columns()取列标签  
loc[]#显式索引， iloc[], at[]#按行列标签访问元素, iat[], T#转置

```python
print(df.index)
print(df.columns)
print(df.values)
print(df.ndim)
print(df.dtypes)
print(df.shape)
print(df.size)
print(df.T, df.T.index)
```

```python
print(df.loc[3])
print(df.iloc[2])
print(df.loc[:,"name"])  #取列
print(df.iloc[:2,0])
print(df.at[3, "name"])
print(df.iat[1, 1])
```

```python
print(df,"\n")

print(df['name'])
print(df.name)
print(type(df.name), type(df['name']))
```

```python
x = df[['name']]
print(x)
print(type(x))

y = df[['name', 'age']]
print(y)
print(type(y))
```

```python
print(df.head(2))
print(df.tail(2))
```

```python
#布尔索引
print(df[df.age > 10])
print(df[(df['age'] > 10) & (df.id < 3)])
```

```python
#随机抽样
print(df.sample(2))
```
