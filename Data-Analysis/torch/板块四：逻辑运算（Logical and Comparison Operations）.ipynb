# 板块四：逻辑运算（Logical and Comparison Operations）

## 一、比较运算（返回布尔张量）
所有比较操作返回 torch.BoolTensor（元素为 True/False）。

| 操作     | 函数形式               | 示例     |
|----------|------------------------|----------|
| 等于     | `==` 或 `torch.eq()`   | `x == y` |
| 不等于   | `!=` 或 `torch.ne()`   | `x != y` |
| 大于     | `>` 或 `torch.gt()`    | `x > 0`  |
| 小于     | `<` 或 `torch.lt()`    | `x < 1`  |
| 大于等于 | `>=` 或 `torch.ge()`   | `x >= 0.5` |
| 小于等于 | `<=` 或 `torch.le()`   | `x <= 10` |

```python
import torch

x = torch.tensor([1, 2, 3, 4, 5], dtype=torch.float32)
y = torch.tensor([1, 3, 3, 5, 4], dtype=torch.float32)

print("x == y :", x == y)   # tensor([True, False, True, False, False])
print("x > 3  :", x > 3)    # tensor([False, False, False, True, True])
print("x <= y :", x <= y)   # tensor([True, True, True, True, False])

# 函数形式（等价）
print("torch.eq(x, y):", torch.eq(x, y))
```

## 二、逻辑运算（布尔张量之间的操作）
对布尔张量进行逻辑组合。

| 操作   | 符号 | 函数                   |
|--------|------|------------------------|
| 逻辑与 | `&`  | `torch.logical_and()`  |
| 逻辑或 | `|`  | `torch.logical_or()`   |
| 逻辑非 | `~`  | `torch.logical_not()`  |

```python
x = torch.tensor([1, 2, 3, 4, 5], dtype=torch.float32)

# 找出 2 < x <= 4 的元素
mask = (x > 2) & (x <= 4)
print("Mask:", mask)                # [False, False, True, True, False]
print("Filtered x:", x[mask])       # [3., 4.]

# 逻辑或：x <= 2 或 x >= 4
mask2 = (x <= 2) | (x >= 4)
print("Or mask:", mask2)            # [True, True, False, True, True]

# 逻辑非
not_mask = ~mask
print("Not mask:", not_mask)        # [True, True, False, False, True]
```

## 三、布尔张量的应用：高级索引（Boolean Indexing）

### 示例：数据清洗

```python
scores = torch.tensor([85., 92., 45., 78., 96.])
passing = scores >= 60
print("Passing scores:", scores[passing])  # [85., 92., 78., 96.]
```

### 更复杂的例子

```python
# 练习：找出张量中绝对值大于 2 的元素，并将其置为 0
data = torch.tensor([-3., -1., 0., 2., 4.])
mask = data.abs() > 2
result = torch.where(mask, torch.tensor(0.), data)
print("After masking:", result)  # [0., -1., 0., 2., 0.]
```

## 四、其他实用函数

| 函数                              | 作用                                               |
|-----------------------------------|----------------------------------------------------|
| `torch.where(condition, x, y)`    | 条件选择：condition 为真取 x，否则取 y             |
| `torch.nonzero(mask)`             | 返回 mask 中 True 元素的索引                       |
| `mask.any()` / `mask.all()`       | 判断是否有任意/全部为 True                         |

```python
x = torch.tensor([-1., 2., -3., 4.])
# 将负数替换为 0（ReLU 效果）
relu_x = torch.where(x > 0, x, torch.tensor(0.0))
print("ReLU via where:", relu_x)  # [0., 2., 0., 4.]

# 等价于：
relu_x2 = x.clamp(min=0)
```

# 总结

| 要点             | 说明                                      |

| 布尔索引         | `x[mask]` 可高效筛选数据                  |
| `torch.where`    | 实现条件赋值                              |
| 注意力掩码       | NLP 中屏蔽 padding 的标准做法             |
