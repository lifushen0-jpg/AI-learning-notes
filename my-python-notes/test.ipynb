```python
def re_read(s:str):
        n = len(s)
        e = ""
        for i in range(n):
            e += s[-i-1]
        return e

re_read("abcd")
```

```python
from PIL import Image
import os

image_folder = "photos"          
output_path = "merged_image.jpg" 
cols = 5                         
rows = 3                         
thumbnail_size = (200, 400)      

extensions = ('.png', '.jpg', '.jpeg', '.bmp', '.gif')
image_files = [os.path.join(image_folder, f) for f in sorted(os.listdir(image_folder)) if f.lower().endswith(extensions)]

image_files = image_files[:15]

if len(image_files) < 15:
    raise ValueError(f"需要15张图片，但只找到 {len(image_files)} 张")

images = []
for img_path in image_files:
    img = Image.open(img_path)
    img = img.resize(thumbnail_size, Image.Resampling.LANCZOS)  
    images.append(img)

merged_width = cols * thumbnail_size[0]
merged_height = rows * thumbnail_size[1]
merged_image = Image.new('RGB', (merged_width, merged_height), color='white')

for i, img in enumerate(images):
    x = (i % cols) * thumbnail_size[0]
    y = (i // cols) * thumbnail_size[1]
    merged_image.paste(img, (x, y))

merged_image.save(output_path)
print(f"保存至 {output_path}")
```
