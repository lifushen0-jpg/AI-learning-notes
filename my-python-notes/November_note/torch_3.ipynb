### 张量的使用

```python
import torch

# 创建张量
x = torch.tensor([1, 2, 3])                # 从列表创建
y = torch.zeros(2, 3)                      # 全零张量
z = torch.randn(3, 4)                      # 标准正态分布随机张量
a = torch.arange(0, 10, 2)                 # [0, 2, 4, 6, 8]

# 查看属性
print("Shape:", x.shape)        # torch.Size([3])
print("Device:", x.device)      # cpu (默认)
print("dtype:", x.dtype)        # torch.int64

# 移动到 GPU（如果可用）
if torch.cuda.is_available():
    x = x.cuda()  # 或 x = x.to('cuda')
```

## 2. 张量的矩阵运算

```python
A = torch.tensor([[1., 2.], [3., 4.]])
B = torch.tensor([[5., 6.], [7., 8.]])

# 矩阵乘法
C = torch.mm(A, B)          # 或 A @ B
print("A @ B =\n", C)

# 转置
A_T = A.t()                 # 或 A.transpose(0, 1)
print("A^T =\n", A_T)

# 批量矩阵乘法: (b, n, m) @ (b, m, p) -> (b, n, p)
batch_A = torch.randn(2, 3, 4)
batch_B = torch.randn(2, 4, 5)
batch_C = torch.bmm(batch_A, batch_B)  # shape: (2, 3, 5)
print("Batch matmul shape:", batch_C.shape)
```

## 3.聚合运算

```python
x = torch.tensor([[1., 2., 3.], [4., 5., 6.]])

print("Sum all:", x.sum())               # 21.0
print("Sum along dim=0:", x.sum(dim=0))  # [5., 7., 9.]
print("Mean along dim=1:", x.mean(dim=1))# [2., 5.]

max_val, max_idx = x.max(dim=1)  # 返回值和索引
print("Max values:", max_val)    # [3., 6.]
print("Argmax:", max_idx)        # [2, 2]
```

## 4.逻辑运算

```python
x = torch.tensor([1, 2, 3, 4])
y = torch.tensor([1, 3, 3, 5])

print("x > 2:", x > 2)           # tensor([False, False, True, True])
print("x == y:", x == y)         # tensor([True, False, True, False])
print("x >= y & x <= 4:", (x >= y) & (x <= 4))  # 逻辑与

# 使用掩码筛选
mask = x > 2
print("Filtered:", x[mask])      # tensor([3, 4])
```
