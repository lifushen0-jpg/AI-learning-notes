# 迭代器与生成器

# Python迭代器（Iterator）

## 1. 迭代器是什么

迭代器（Iterator）是 Python 中一种**能够记住当前位置，并且逐个返回元素的对象**。

简单理解：

> 迭代器不是数据本身，而是访问数据的一种工具。

一个迭代器包含：

- 当前访问位置
- 获取下一个元素的方法   


例如：

````python
numbers = [10, 20, 30, 40]

it = iter(numbers)

print(next(it))
print(next(it))
````

其中：

- `numbers` 是数据源
- `it` 是迭代器
- `next(it)` 获取下一个元素


# 2. 迭代器的工作过程

例如：

````python
numbers = [10, 20, 30, 40]

it = iter(numbers)
print(next(it))
print(next(it))
print(next(it))
````




创建迭代器后：

```
数据:
+----+----+----+----+
|10  |20  |30  |40  |
+----+----+----+----+

迭代器位置:
 ^
```

第一次调用：

```python
next(it)
```

返回：

```text
10
```

迭代器移动：

```
数据:
+----+----+----+----+
|10  |20  |30  |40  |
+----+----+----+----+
      ^
```

第二次：

```python
next(it)
```

返回：

```text
20
```

因此：

> 迭代器本质上是一个带状态的数据访问对象。


# 3. 可迭代对象（Iterable）

如果一个对象具有：

```python
__iter__()
```

方法，那么它是可迭代对象。

常见可迭代对象：

|对象|是否可迭代|
|-|-|
|list|✓|
|tuple|✓|
|string|✓|
|dict|✓|
|set|✓|
|file|✓|
|generator|✓|

例如：

```python
lst = [1,2,3]

it = iter(lst)
```

其中：

- `lst` 是可迭代对象
- `it` 是迭代器


# 4. 迭代器协议

一个对象成为迭代器，需要实现：

```python
__iter__()
```

和：

```python
__next__()
```

两个方法。


## 4.1 __iter__()

作用：

返回迭代器对象。

等价：

```python
iter(obj)
```

等价于：

```python
obj.__iter__()
```


## 4.2 __next__()

作用：

返回迭代器中的下一个元素。

等价：

```python
next(it)
```

等价于：

```python
it.__next__()
```


# 5. iter()函数

## 作用

将可迭代对象转换为迭代器。

语法：

```python
iter(object)
```

例如：

```python
lst = [1,2,3]

it = iter(lst)

print(type(it))
```

输出：

```text
<class 'list_iterator'>
```

内部：

```python
iter(lst)
```

等价于：

```python
lst.__iter__()
```


# 6. next()函数

## 作用

从迭代器中获取下一个元素。

语法：

```python
next(iterator)
```

例如：

```python
it = iter([1,2,3])

print(next(it))
print(next(it))
print(next(it))
```

输出：

```text
1
2
3
```


## 6.1 StopIteration异常

当迭代器没有更多元素：

```python
next(it)
```

会产生：

```text
StopIteration
```

异常。


例如：

```python
it = iter([1])

print(next(it))
print(next(it))
```

结果：

```text
1

StopIteration
```


## 6.2 next()默认值

语法：

```python
next(iterator, default)
```

例如：

```python
it = iter([1])

print(next(it,"结束"))
print(next(it,"结束"))
```

输出：

```text
1
结束
```


# 7. for循环的本质

Python：

```python
for x in iterable:
    print(x)
```

实际上等价于：

```python
it = iter(iterable)

while True:
    try:
        x = next(it)
        print(x)
    except StopIteration:
        break
```

因此：

```
for循环
    |
    ↓
调用iter()
    |
    ↓
得到迭代器
    |
    ↓
不断调用next()
    |
    ↓
遇到StopIteration结束
```


# 8. 迭代器和列表的区别


## 8.1 列表

列表一次性保存所有数据：

```python
lst = [1,2,3,4,5]
```

内存：

```
[1,2,3,4,5]
```

所有元素同时存在。


## 8.2 迭代器

迭代器不会复制数据。

它只保存：

```
当前位置
+
获取下一个元素的方法
```

例如：

```python
it = iter(lst)
```

类似：

```
数据:
[1,2,3,4,5]

迭代器:
位置 = 0
```


# 9. 自定义迭代器

可以通过类实现迭代器：

```python
class Counter:
    def __init__(self,n):
        self.n = n
        self.i = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.i < self.n:
            self.i += 1
            return self.i
        else:
            raise StopIteration


c = Counter(3)

print(next(c))
print(next(c))
print(next(c))
```

输出：

```text
1
2
3
```


# 10. 迭代器与生成器

生成器（Generator）是一种特殊的迭代器。

关系：

```
可迭代对象 Iterable
        |
        | iter()
        ↓
迭代器 Iterator
        |
        ↓
生成器 Generator
```

更准确：

```
Generator ⊂ Iterator ⊂ Iterable
```

即：

- 所有生成器都是迭代器
- 所有迭代器都是可迭代对象
- 普通列表不是迭代器


例如：

```python
lst = [1,2,3]

next(lst)
```

错误：

```text
TypeError
```

需要：

```python
next(iter(lst))
```


# 11. yield与迭代器

使用 `yield` 的函数会自动创建生成器。

例如：

```python
def func():
    yield 1
    yield 2
    yield 3


g = func()

print(next(g))
print(next(g))
```

输出：

```text
1
2
```


生成器执行过程：

```
调用函数
    |
    ↓
创建generator对象
    |
    ↓
next()
    |
    ↓
执行到yield
    |
    ↓
返回值
    |
    ↓
保存当前状态
    |
    ↓
下一次next继续执行
```


# 12. 迭代器的意义


## 12.1 统一访问方式

不同数据结构：

```
list
tuple
dict
set
string
file
generator
```

都可以：

```python
iter()
```

转换为迭代器。

然后：

```python
next()
```

访问。


## 12.2 节省内存

迭代器采用：

> 惰性计算（Lazy Evaluation）

不会提前生成所有数据。


例如：

普通列表：

```python
numbers = [0,1,2,...,999999]
```

需要一次性保存所有数字。


迭代器：

```python
next(iterator)
```

需要一个，生成一个。


# 13. 迭代器的应用


## 13.1 文件读取

普通方式：

```python
data = open("file.txt").readlines()
```

一次读取所有内容。


迭代方式：

```python
with open("file.txt") as f:
    for line in f:
        print(line)
```

文件对象本身就是迭代器。


## 13.2 PyTorch DataLoader

训练模型：

```python
for batch in dataloader:
    train(batch)
```

本质：

```python
batch = next(iterator)
```

不断产生训练数据。


## 13.3 无限序列

例如：

```python
def count():
    i = 0
    while True:
        yield i
        i += 1


c = count()

print(next(c))
print(next(c))
print(next(c))
```

输出：

```text
0
1
2
```

虽然序列无限：

```
0,1,2,3,4,...
```

但是不会占用无限内存。


# 14. 总结


## 核心概念

|概念|含义|
|-|-|
|Iterable（可迭代对象）|可以使用iter()转换的对象|
|Iterator（迭代器）|保存当前位置并逐个返回元素的对象|
|Generator（生成器）|使用yield创建的特殊迭代器|


## 核心函数

|函数|作用|
|-|-|
|iter()|创建迭代器|
|next()|获取下一个元素|


## 核心流程

```
可迭代对象
      |
      | iter()
      ↓
迭代器
      |
      | next()
      ↓
元素
```


## 最核心理解

> 迭代器就是一个保存状态、能够逐步产生数据的对象。

它解决的问题不是存储数据，而是：

> 如何高效、统一、逐个访问数据。
