## 创建array

2. 使用arange( , ,)，语法同python(略)

3. 使用linspace(起始值，终止值，步长)生成等差数列

```python
array1 = np.linspace(1,14,5)
print(array1)
```

4. 使用logspace(起始指数，终止指数，num=元素个数，base=底数)生成等比数列

```python
array2 = np.logspace(1,14,num=5,base = 1.1)
print(array2)
```
