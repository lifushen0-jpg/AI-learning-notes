# 散点图

```python
import torch
```

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

# 创建图表并设置大小
plt.figure(figsize = (10,5))

# ====================================================================================================
# ====================================================================================================

# 图表数据
x = torch.rand(1,500)*2
y = x*6 + torch.rand_like(x)*6

plt.scatter(x,y,
            alpha=0.5),
plt.xlabel('subject',fontsize = 10)
plt.ylabel('score', fontsize = 10)

plt.grid(axis="y", alpha = 0.5, color = 'blue', linestyle = '--')  #plt.grid(axis = 'x')  # 


# ====================================================================================================
# ====================================================================================================

# 自动优化排版(效果较差)
plt.tight_layout()

# 添加标题
plt.title("guass noisy",color = 'red', fontsize = 20)

# 显示图表
plt.show()
```

# 箱线图

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
rcParams['font.family'] = 'SimHei'

# 创建图表并设置大小
plt.figure(figsize = (10,5))

# ====================================================================================================
# ====================================================================================================

data_dict ={
    'Chinese' : (torch.rand(20)*30+70).numpy(),
    'Math': (torch.rand(20)*30+70).numpy(),
    'English':(torch.rand(20)*30+70).numpy(),
}

# 提取值（必须是 list of 1D arrays）
data = list(data_dict.values())
labels = list(data_dict.keys())
plt.boxplot(data, labels= labels) # 只能接受一维array数据

plt.xlabel('Variables')
plt.ylabel('Values')


# ====================================================================================================
# ====================================================================================================

# 自动优化排版(效果较差)
plt.tight_layout()

# 添加标题
plt.title("guass noisy",color = 'red', fontsize = 20)

# 显示图表
plt.show()
```
