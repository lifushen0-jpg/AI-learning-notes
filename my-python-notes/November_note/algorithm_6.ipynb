# 二叉树（Binary Tree）  
—— 计算机科学中最基础、最重要的树形数据结构之一

---

## 一、定义

**二叉树** 是一种每个节点最多有两个子节点的树结构，这两个子节点分别称为：

- **左子节点**（left child）  
- **右子节点**（right child）

并且，**左右子树是有顺序的**：即使某个节点只有一个子节点，也必须明确它是左孩子还是右孩子。

### 形式化定义（递归）：
- 一棵二叉树可以是 **空树**（没有任何节点）；
- 或者由一个**根节点**和两棵互不相交的、分别称为**左子树**和**右子树**的二叉树组成。

---

## 二、基本术语

| 术语 | 说明 |
|------|------|
| **根**（Root） | 树的起始节点，没有父节点 |
| **叶子**（Leaf） | 没有子节点的节点 |
| **内部节点** | 至少有一个子节点的非根节点 |
| **深度**（Depth） | 从根到该节点的边数（根深度为 0） |
| **高度**（Height） | 从该节点到最远叶子的最长路径边数（叶子高度为 0） |
| **层**（Level） | 根在第 0 层或第 1 层（依定义而定），其子节点在下一层 |

> 📌 注意：树的高度 = 根节点的高度；树的深度通常指最大节点深度，等于高度。

---

## 三、二叉树 vs 普通树

| 特性 | 二叉树 | 普通树 |
|------|--------|--------|
| 子节点数量 | ≤ 2 | 任意多个 |
| 子节点顺序 | 有左右之分（有序） | 通常无序（除非特别说明） |
| 空子树 | 允许左或右为空 | 一般只关心是否有孩子 |

> ⚠️ **注意**：二叉树不是“度 ≤ 2 的树”那么简单——**左右位置不能互换**。例如下面两棵树不同：

```
    A           A
   /             \
  B               B
```

尽管结构对称，但在二叉树中它们是**不同的树**。

---

## 四、常见类型

1. **满二叉树**（Full Binary Tree）  
   - 每个节点要么是叶子，要么有两个子节点。

2. **完全二叉树**（Complete Binary Tree）  
   - 除最后一层外全满，最后一层节点靠左排列（可用于数组存储）。

3. **完美二叉树**（Perfect Binary Tree）  
   - 所有叶子在同一层，且所有非叶节点都有两个子节点。  
   - 节点总数为 $2^{h+1} - 1$，其中 $h$ 为树的高度。

4. **二叉搜索树**（BST, Binary Search Tree）  
   - 对任意节点：左子树所有节点值 < 根值 < 右子树所有节点值。  
   - 支持高效的查找、插入、删除（平均 $O(\log n)$）。

5. **平衡二叉树**（如 AVL 树、红黑树）  
   - 左右子树高度差不超过 1（AVL）或满足颜色规则（红黑树）。  
   - 保证最坏情况下的操作效率仍为 $O(\log n)$。

---

## 五、存储方式

### 1. 链式存储（常用）

每个节点包含：
- 数据域（`data`）
- 左孩子指针（`left`）
- 右孩子指针（`right`）

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

### 2. 顺序存储（仅适合完全二叉树）

用数组表示，利用下标关系（以 0 为起始索引）：

- 父节点 `i` → 左孩子 `2i + 1`，右孩子 `2i + 2`
- 孩子 `i` → 父节点 `(i - 1) // 2`

>  优点：节省指针空间，缓存友好  
>  缺点：稀疏树浪费大量空间

---

## 六、遍历方式（核心操作）

| 遍历类型 | 顺序 | 应用场景 |
|----------|------|----------|
| **前序遍历** | 根 → 左 → 右 | 复制树、打印目录结构 |
| **中序遍历** | 左 → 根 → 右 | BST 升序输出 |
| **后序遍历** | 左 → 右 → 根 | 删除树、计算目录大小 |
| **层序遍历** | 从上到下、从左到右 | 广度优先搜索（BFS） |

### 递归实现示例（Python）

````python
def preorder(root):
    if root:
        print(root.val)
        preorder(root.left)
        preorder(root.right)

def inorder(root):
    if root:
        inorder(root.left)
        print(root.val)
        inorder(root.right)

def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.val)


### 层序遍历（使用队列）

from collections import deque

def level_order(root):
    if not root:
        return
    q = deque([root])
    while q:
        node = q.popleft()
        print(node.val)
        if node.left:
            q.append(node.left)
        if node.right:
            q.append(node.right)
````

# 🌟 堆排序（Heap Sort）详解  
—— 基于堆的高效原地排序算法

---

## 一、什么是堆排序？

**堆排序**是一种基于“堆”数据结构的**比较型、原地、不稳定**排序算法。其核心思想是：

> 利用最大堆（或最小堆）的性质，反复取出当前最大（或最小）元素，放到数组末尾，从而完成排序。

- 使用 **最大堆** → 实现 **升序排序**
- 使用 最小堆 → 理论上可实现降序，但因需额外空间或反向操作，**通常不采用**

堆排序的时间复杂度稳定为 $O(n \log n)$，且仅需 $O(1)$ 额外空间，是理论性能优秀的排序算法之一。

---

## 二、关键概念：什么是“堆”？

**堆 = 完全二叉树 + 堆性质**

### 1. 完全二叉树
- 除最后一层外，其他层全部填满；
- 最后一层节点**靠左排列**。

### 2. 堆性质
- **最大堆**：任意父节点 ≥ 其子节点 ⇒ 根为全局最大值  
- **最小堆**：任意父节点 ≤ 其子节点 ⇒ 根为全局最小值

### ✅ 数组表示堆（无需指针）

对于索引为 `i` 的节点（从 0 开始）：
- 左孩子：`2*i + 1`
- 右孩子：`2*i + 2`
- 父节点：`(i - 1) // 2`

> **示例**：数组 `[50, 30, 40, 10, 20]` 对应的最大堆结构为：

```
       50
      /  \
    30    40
   /  \
 10   20
```

---

## 🔁 三、堆排序的两个核心步骤

### 步骤 1️⃣：构建最大堆（Build Max Heap）

将无序数组原地调整为一个最大堆。

- **方法**：从**最后一个非叶子节点**开始，自底向上执行 `heapify`（堆化）
- **最后一个非叶子节点索引**：$\left\lfloor \frac{n}{2} \right\rfloor - 1$

> 为什么从这里开始？因为叶子节点天然满足堆性质，无需调整。

---

### 步骤 2️⃣：排序（Repeatedly Extract Max）

重复以下操作 $n - 1$ 次：

1. 将堆顶（当前最大值）与**堆的最后一个元素**交换；
2. **缩小堆的大小**（即排除已排好序的尾部元素）；
3. 对新的堆顶执行 `heapify`，恢复最大堆性质。

最终，整个数组变为**升序排列**。

---

## 🧩 四、直观例子演示

**原始数组**：`[4, 10, 3, 5, 1]`

### 第一步：建最大堆

初始数组：  
`[4, 10, 3, 5, 1]`

从索引 `(5//2)-1 = 1` 开始 heapify：

- 对索引 1（值=10）：已是堆
- 对索引 0（值=4）：左右子为 10 和 3 → 与 10 交换 → `[10, 4, 3, 5, 1]`
- 再对子树（索引 1）heapify：4 的子为 5 和 1 → 与 5 交换 → `[10, 5, 3, 4, 1]`

✅ **建堆后**：`[10, 5, 3, 4, 1]`

### 第二步：反复取最大值

| 操作 | 数组状态 | 说明 |
|------|----------|------|
| 交换 10 ↔ 1，heapify 前4个 | `[1, 5, 3, 4, 10]` → heapify → `[5, 4, 3, 1, 10]` | 最大值 10 归位 |
| 交换 5 ↔ 1，heapify 前3个 | `[1, 4, 3, 5, 10]` → heapify → `[4, 1, 3, 5, 10]` | 5 归位 |
| 交换 4 ↔ 3，heapify 前2个 | `[3, 1, 4, 5, 10]` → heapify → `[3, 1, 4, 5, 10]` | 4 归位 |
| 交换 3 ↔ 1 | `[1, 3, 4, 5, 10]` | 3 和 1 自动有序 |

✅ **最终结果**：`[1, 3, 4, 5, 10]` —— 升序！

````python
def heapify(arr, n, i):
    """
    对以索引 i 为根的子树进行堆化（维护最大堆性质）
    
    参数:
        arr: 待排序的数组
        n: 当前堆的有效大小（不是整个数组长度！）
        i: 当前要堆化的节点索引（根）
    """
    largest = i          # 假设当前根是最大值
    left = 2 * i + 1     # 左孩子索引
    right = 2 * i + 2    # 右孩子索引

    # 如果左孩子存在且比当前最大值还大
    if left < n and arr[left] > arr[largest]:
        largest = left

    # 如果右孩子存在且比当前最大值还大
    if right < n and arr[right] > arr[largest]:
        largest = right

    # 如果最大值不是根节点，则需要交换并继续向下堆化
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]  # 交换
        heapify(arr, n, largest)  # 递归堆化被影响的子树


def heap_sort(arr):
    """
    堆排序主函数（升序）
    
    步骤:
        1. 构建最大堆
        2. 反复将堆顶（最大值）移到末尾，并缩小堆范围
    """
    n = len(arr)

    # 🚧 第一步：构建最大堆
    # 从最后一个非叶子节点开始，逆序向上堆化
    # 最后一个非叶子节点索引 = n//2 - 1
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # 🔁 第二步：逐个提取最大值（放到数组末尾）
    for i in range(n - 1, 0, -1):
        # 将当前最大值（堆顶）与末尾元素交换
        arr[0], arr[i] = arr[i], arr[0]
        
        # 对剩余的 i 个元素（0 到 i-1）重新堆化（堆大小变为 i）
        heapify(arr, i, 0)


# ======================
# 🧪 测试示例
# ======================
if __name__ == "__main__":
    data = [12, 11, 13, 5, 6, 7]
    print("原始数组:", data)
    
    heap_sort(data)
    
    print("排序后数组:", data)
    # 输出: [5, 6, 7, 11, 12, 13]
````

## 时间复杂度

- ### 1. 构建堆：O(n)
- ### 2. 维护大顶堆：O(log(n))  堆的高度为log(n)
