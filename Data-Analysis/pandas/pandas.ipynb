## 第一步：Pandas基础 - Series和DataFrame

```python
# 创建Series的几种方式

# 从列表创建
s1 = pd.Series([10, 20, 30, 40])
print("从列表创建：")
print(s1)

# 从字典创建（自动使用键作为索引）
s2 = pd.Series({'苹果': 5, '香蕉': 3, '橙子': 8})
print("\n从字典创建：")
print(s2)

# 自定义索引
s3 = pd.Series([100, 200, 300], index=['A', 'B', 'C'])
print("\n自定义索引：")
print(s3)
```

```python
temperature = pd.Series(
    [25, 28, 26, 30, 27],
    index=["周一","周二","周三","周四","周五"])

print(temperature)
print(f"本周三的温度为：{temperature['周三']}")  # 可以用这个方法查询标签对应元素

# fruits = {'苹果': 5, '香蕉': 3, '橙子': 8}
# print(f"apple:{fruits["苹果"]}")
```

```python
# 创建DataFrame的几种方式
import pandas as pd

# 方式1：从字典创建（最常用）
# 这种方法会把键名当作列名创建
# 字典的值必须是列表

data = {
    '姓名': ['张三', '李四', '王五'],
    '数学': [85, 92, 78],
    '语文': [88, 90, 85],
    '英语': [92, 88, 80]
}

df = pd.DataFrame(data)
print("学生成绩表：")
print(df)
print("\nDataFrame的形状（行数, 列数）：", df.shape)
print("列名：", df.columns.tolist())
print("索引：", df.index.tolist())

# 方式2：从列表的列表创建

data_list = [
    ['张三', 85, 88, 92],
    ['李四', 92, 90, 88],
    ['王五', 78, 85, 80]
]

df2 = pd.DataFrame(data_list, columns=['姓名', '数学', '语文', '英语'])
print("\n从列表创建：")
print(df2)
```

# 练手项目：学生管理系统

项目目标：
创建一个学生信息管理系统，包含以下功能：

1. 创建学生信息表  
2. 添加新学生  
3. 查询学生信息  
4. 计算平均分  

```python
import pandas as pd

# 1. 创建初始学生表
students = pd.DataFrame({
    '学号': ['S001', 'S002', 'S003'],
    '姓名': ['张三', '李四', '王五'],
    '数学': [85, 92, 78],
    '英语': [88, 90, 85],
    '物理': [90, 85, 88]
})

print("初始学生表：")
print(students)
print("\n" + "="*50)
```

```python
zhaoliu = pd.DataFrame({
    "学号": ["S004"],
    "姓名": ["赵六"],
    "数学": [95],
    "英语": [88],         
    "物理": [92]
})

students = pd.concat([students, zhaoliu], axis=0, ignore_index=True)
print(students)
```

```python
# 方法1：使用列名（最简单）
math_scores = students['数学']
print("所有学生的数学成绩：")
print(math_scores)

# 方法2：使用点号（列名没有空格时可用）
math_scores = students.数学
print("\n使用点号访问：")
print(math_scores)

# 方法3：选择多列
scores = students[['姓名', '数学', '英语']]
print("\n姓名和成绩：")
print(scores)
```

```python
# 方法1：布尔索引（最常用）
high_math = students[students['数学'] > 90]
print("数学高于90分的学生：")
print(high_math)

# 方法2：使用query()方法（类似SQL）
high_math = students.query('数学 > 90')
print("\n使用query方法：")
print(high_math)

# 方法3：多个条件（AND操作）
# 数学>90 AND 英语>85
top_students = students[(students['数学'] > 90) & (students['英语'] > 85)]
print("\n数学>90且英语>85的学生：")
print(top_students)

# 方法4：多个条件（OR操作）
# 数学>90 OR 英语>90
good_students = students[(students['数学'] > 90) | (students['英语'] > 90)]
print("\n数学>90或英语>90的学生：")
print(good_students)
```
