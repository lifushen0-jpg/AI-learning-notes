```python
import pandas as pd
import numpy as np
```

数据类型

```python
#生成时最前面一列为索引
#一维数据（列表）（数据类型相同）
print(pd.Series([1,2,3]))
#二维数据（字典）（各列数据类型相同）
print(pd.DataFrame({'col':[1,2,3]}))
```

Series:

```python
s = pd.Series([10,3,4,93,4], index = ["A", "B", "C", "D", "E"])
print(s)

s = pd.Series([1, 2, 4, 5, 6], name = "month")
print(s)

s = pd.Series({ "a": 1, "b": 2, "c": 3 })
print(s)
print(pd.Series(s, index = ["a", "c"]))
```

```python
s1 = pd.Series([1,2,3,4,5,6], index = ["A", "B", "C", "D", "E", "F"])
s2 = pd.Series({ "a": 1, "b": 2, "c": 3 })
print(s1.index)

print(s1.values)
print(s2.values)

print(s1.shape, s2.shape,s1.ndim, s1.size)

print(s1.loc["B"])
print(s1.iloc[2])
print(s1.loc["A":"D"])
print(s1.iloc[1:3])

print(s1.at["B"], s1.iat[2])
```

访问数据

```python
print(s1[s1 < 3])
print(s1.head())   #返回前五行
print(s1.tail())   #返回后五行
```

Series中的常用函数

```python
s = pd.Series([10, 2, np.nan ,None, 3, 4, 5], index = ["A", "B", "C", "D", "E", "F", "G"])
print(s)
```

```python
print(s.describe())  #获取所有描述性信息，包括：count(忽略None),mean,std,min,max,四分位数，
print(s.var())      #方差
print(s.median())   #中位数
print(s.quantile(0.1))  #分位数
print(s.mode())   #众数
print(s.value_counts())   #记录元素出现频率

print(s.keys())
print(s.index)
```

```python
print(s.isna())   #检查是否为缺失值

print(s.isin([4,10,9]))  #检查是否在给定列表中

print(s.drop_duplicates())  # 去重
```

```python
print(s.sort_index())
print(s.sort_values())
```

实战：

```python
temperatures = pd.Series([28, 31, 29, 32, 30, 27, 33],
                         index= ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun'])

n = temperatures[temperatures > 30].count()
print(n)

t2 = temperatures.sort_values(ascending= False)
print(t2)

mean = temperatures.mean()
print(mean)

#获取相邻两个数据之间的差值
diff_t = temperatures.diff().abs()
t3 = diff_t.sort_values(ascending= False).index
print(t3[:2].tolist())  # 使用tolist转化为列表输出
```
