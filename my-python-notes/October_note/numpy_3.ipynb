```python
import numpy as np
```

```python
x = np.array([[5, 0, 3, 3],
              [7, 9, 3, 5],
              [2, 4, 7, 6]])
print(x)
print(np.count_nonzero(x < 6))
print(np.sum(x < 6))
print(np.sum(x < 6, axis=1))

print(np.any(x > 8))
print(np.all(x < 10))
```
