```python
import torch
import numpy as np
import pandas as pd
import os
```

```python
# 用GPU运行
dev = torch.device('cuda')

a = torch.tensor([2,2],
                 device = dev)
print(a)
```

```python
# 稀疏的张量
# 稀疏张量（Sparse Tensor）是 PyTorch 中一种高效存储和计算大量零值数据的数据结构。
# 它的核心思想是：只存储非零元素及其位置，忽略零值，从而节省内存并加速某些运算。
i = torch.tensor([[0,1,2],[0,1,2]])
v = torch.tensor([1,2,3])
x = torch.sparse_coo_tensor(i, v, (4,4))
print(x)
print(x.to_dense())
```

### 加减乘（哈达玛及积）除幂


```python
a = torch.tensor([[4,5,6],[7,8,9]])
b = torch.tensor([[1,2,3],[4,5,6]])

print(a)
print(b)
```

```python
print(a + b)
print(a.add(b))
print(torch.add(a,b))
print(a)
print(a.add_(b))  #将 a 的值更改为加完后的结果
print(a)
```

```python
print(a.sub_(b))
```

```python
print(a * b)
print(torch.mul(a,b))
```

```python
print(a / b)
print(torch.div(a,b))
```

```python
print(a)
print(a ** 3)
print(a.pow(3))
```

```python
print(torch.exp(a))
print(a.exp())
```

```python
print(torch.log(a))  # 以e为底
print(a.log())

print(a.log2())
print(a.log10())
```

```python
print(torch.sqrt(a))
print(a.sqrt())
```

### 矩阵运算

```python
a = torch.ones(2,1)
b = torch.ones(1,2)
print(a @ b)
print(a.matmul(b))
print(torch.matmul(a,b))
```

```python
a = torch.ones(1,2,3,4)
b = torch.ones(1,2,4,3)
print(a @ b)           #高维张量只对最后两个维度进行矩阵乘法
print(a.matmul(b))
print(torch.matmul(a,b))
```

## 广播机制：两个张量都不为零维且右对齐

```python
# 右对齐(从右向左对齐，若对应位置上的元素有一个为1或空， 或两者相等则判定为对齐)
x = torch.rand(3)
y = torch.rand(2,2,1)
z = x + y
print(x)
print(y)
print(z)
print(z.shape)
```

```python
a = torch.rand(2,2)
a = a*10
print(a)

print(torch.floor(a))
# print(torch.floor_(a))
print(torch.round(a))  # 向下取整
print(torch.ceil(a))   # 四舍五入
print(torch.trunc(a))  # 向上取整
print(torch.frac(a))   # 取小数
print(a % 2)           # 取余
```

```python
a = torch.tensor([[1,2,3], [2,5,3]])
b = torch.tensor([[1,4,3], [2,6,3]])
print(a, "\n\n", b)

print(torch.eq(a, b))      # 对应元素是否相等
print(torch.equal(a, b))   # 两个张量是否相等
```

```python
print(torch.ge(a,b))
print(torch.gt(a,b))
print(torch.le(a,b))
print(torch.lt(a,b))
print(torch.ne(a,b))
```

```python
print(a)
print(torch.sort(a))
print(a)
print(torch.sort(a,dim=0))
print(torch.sort(a, descending=True))
```

```python
a = torch.tensor([
    [2,3,4,2,5],
    [3,2,4,6,1]
])

print(torch.topk(a, k = 2, dim = 1))
print(torch.topk(a, k = 1, dim = 0))

print(torch.kthvalue(a, k = 2, dim = 1))
```

```python
a = torch.rand(2,3)
print(a)
print(torch.isfinite(a))
print(torch.isfinite(a/0))
print(torch.isinf(a))

b = torch.tensor([1, 2, np.nan])
print(torch.isnan(b))
```
